### filter(), map() and reduce()

### filter()

```python
# - It takes a function and an iterable as arguments and returns a filter object.
# - It filters the values from an iterable.
# - It calls the function on each element of the iterable.
# - For whatever values the fucntion returns True, it stores those values inside the filter object.
# - We can typecast the filter object to list or tuple to get the output.
```

```python
# filter(function,iterable)
```

```python
lst = [54, 12, 35, 61, 14, 25, 31, 94, 31, 22]
res=[]
for i in lst:
    if (lambda x:x%2==0) (i):
        res.append(i)

print(res)
```

```python
f=filter(lambda x:x%2==0, lst)
print(f, type(f))

res=list(f)
print(res)
```

```python
def check_even(x):
    if x%2==0:
        return True
    else:
        return False
```

```python
check_even(4)
```

```python
list(filter(lambda x:x%2==0, lst))
```
