# Functions

```python
# - A function provides us the facility that we write the code once and use it as and when required in a program without writing the code again.
# - A function provides us the facility of code reusability.

# - A function is a block of code that gets executed as and when it is called.
```

# Types of Functions

```python
# - 1- Built-in functions
# - 2- User defined functions
```

### 1- Built-in functions

```python
# - The functions that are predefined in Python are the built-in functions.
# - We can not see the code behind those functions but we can use them.

# Examples- input(), print(), len(), type(), id(), ord(), dir(), strip(), index().....etc.
```

### 2- User defined functions

```python
# - The functions that are defined by the programmer are the user defined functions.
# - def keyword is used to define a user defined function.
```

### Syntax

```python
# def function_name():
#     code
#     code
```

### Components of a function

```python
# - 1- Function definition
# - 2- Function call
```

### 1- Function Definition

```python
# - This is defining the function name and logic (code).
# - A function is defined with def keyword.
```

# Parameterised and Non-parameterised functions

### Parameterised Functions

```python
# - The function that takes a parameter(s) is a parameterised function.
```

### Parameter

```python
# - The variables that we define inside the () of the function definition are known as parameters.
```

```python
def test(a,b): # a and b are parameters
    pass
```

### Argument

```python
# - The values or variables that we define inside the () of the function call are known as arguments.
```

```python
test(5,6) # 5 and 6 are the arguments
```

```python
# - Parameters and arguments are used interchangably often.
```

### Non-Parameterised Functions

```python
# - The function that does not take any parameter is a non-parameterised function.
```

```python
def test(): # non-parameterised function
    pass
```

```python
def test(a,b,c): # parametrised function
    pass
```

# Global and Local variables

### Global variables

```python
# - The variables that we define outside a function definition are the global variables.
# - Global variables has the scope outside the function as well as inside the function.
```

### Local variables

```python
# - The variables that are defined inside a function definition are the local variables.
# - The local variables have the scope only within the function which they have been defined inside.
# - Th local variables have no scope outside the function.
```

```python
a=10 # global variable
print(a)
```

```python
a=15 # Global Variable
def test():
    x=55 # local variable
    a=20 # Local variable
    print(x)
    print(a)
x = 220 # Global Variable
```

```python
print(a)
```

```python
print(x)
```

```python
test()
```

# Types of Arguments

```python
# - 1- Required Positional Arguments
# - 2- Default Arguments
# - 3- Keyworded Arguments
# - 4- Variable Length Arguments
# - 5- Keyworded Variable Length Arguments
```

### 1- Required Positional Arguments

```python
# - The number of parameters and the number of arguments should be equal.
# - The values are passed from the arguments to parameters as per the order they are positioned.
# - Order matters, name does not matter.
```

```python
def course(a,b,c):
    print(a,b,c)
```

```python
course('DS', 'DA','BA')
```

```python
course('BA', 'DA', 'MBA', 'BCA')
```

```python
def course(a,b,c):
    print(a)
    print(b)
    print(c)
```

```python
course('BA', 'DA')
```
